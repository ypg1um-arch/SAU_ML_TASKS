```python
# 1.Add scalar to array
import numpy as np
a = np.array([10, 20, 30])
r = a + 5 #add 5 to each element
print(r)
print('\n')
b = np.array([[1, 2],[3, 4]])
r = b + 10
print(r)
print('\n')
a = np.array([1, 2, 3])
r = np.add(a, 67) # Identical to a + 67
print(r)
```

```python
# 2.Operations between arrays of different shapes
import numpy as np
matrix = np.array([[10, 20, 30],[40, 50, 60]])
scalar = 5
result = matrix + scalar #5 added to each element
print("Scalar Broadcast:\n", result)
print('\n')
matrix = np.array([[1, 2], [3, 4], [5, 6]])
row_vector = np.array([10, 20])
result = matrix + row_vector # row vector added to all rows
print("\nRow-wise Broadcast:\n", result)
print('\n')
matrix = np.array([[1, 2], [3, 4], [5, 6]])
col_vector = np.array([10, 20, 30])[:, None] #this was not written for row as row is default
result = matrix + col_vector # column vector added to all columns
print("\nColumn-wise Broadcast:\n", result)
print('\n')
#code fails for mismatch as it is against the rules mentioned in lecture notes
```

```python
# 3.Create meshgrids for plotting
# need MatPlotLib for plotting results
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-5, 5, 100)  # 100 points between -5 and 5
y = np.linspace(-5, 5, 100)  # same here
X, Y = np.meshgrid(x, y)      # Mesh Grid forms
Z = np.sin(np.sqrt(X**2 + Y**2)) # function for contour map[which will be circular]

# 4. Plot the contour map
plt.contourf(X, Y, Z, cmap='viridis')#use viridis as it prints well in black-and-white
plt.colorbar(label='Z Value')
plt.title("2D Contour Plot using np.meshgrid")
plt.show()

```

```python
# 4.Normalize arrays using broadcasting
import numpy as np
# Min-Max Normalization[all values between 0 and 1]
data = np.array([[10, 100, 0.5],[20, 200, 0.2],[30, 300, 0.8]])
data_min = data.min(axis=0, keepdims=True) #calculate min and max for each column
data_max = data.max(axis=0, keepdims=True)
normalized = (data - data_min) / (data_max - data_min) #normalize
print("Column-Normalized:\n", normalized)
print('\n')
data_min = data.min(axis=1, keepdims=True) #calculate min and max for each row
data_max = data.max(axis=1, keepdims=True)
normalized = (data - data_min) / (data_max - data_min) #normalize
print("Row-Normalized:\n", normalized)
# Z score normalization[for standard normal distribution]
mean = data.mean(axis=1, keepdims=True)  # mean and sd for each row
std  = data.std(axis=1, keepdims=True)
standardized = (data - mean) / std # this is the formula for z score
print("\nRow-Standardized:\n", standardized)
print('\n')
mean = data.mean(axis=1, keepdims=True)  # mean and sd for each column
std  = data.std(axis=1, keepdims=True)
standardized = (data - mean) / std # this is the formula for z score
print("\nColumn-Standardized:\n", standardized)
#theoretical range of z score is -infinity to +infinity, but most values fall in -3 to +3[approximately 99.7% of all data lies in this range]
```
