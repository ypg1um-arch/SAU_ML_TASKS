```python
# 1.Load CSV data and inspect structure
import pandas as pd
df = pd.read_csv('http://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data') #The Iris Dataset(df is a Dataframe)
print("Dataset Shape:", df.shape) #Outputs number of rows and columns
print("\nData Structure Info:")
df.info()#Outputs every column name and other relevant data
print("\nFirst 5 Rows:")
print(df.head()) #Display the first 5 rows
print('\nLast 5 Rows:')
print(df.tail()) #Display the last 5 rows
```

```python
# 2.Handle missing values (dropna, fillna, interpolate)
import pandas as pd
import numpy as np

# Create a dirty dataset with various missing value types[used dictionary]
data = {
    'Date': pd.date_range(start='2026-05-01', periods=6, freq='W'),
    'Store_ID': [101, 101, np.nan, 101, 101, np.nan],
    'Category': ['Electronics', np.nan, 'Electronics', 'Electronics', 'Clothing', np.nan],
    'Temperature_F': [65.0, np.nan, np.nan, 72.0, 75.0, 78.0],
    'Sales_USD': [12000, 14000, 11000, np.nan, 18000, np.nan],
    'Bonus_Payout': [np.nan, np.nan, np.nan, np.nan, np.nan, np.nan]
}

df = pd.DataFrame(data) #data is a dictionary
print("Original DataFrame:")
print(df)
print('\n')
df = df.dropna(axis=1, how='all')#Remove the column that is 100% empty
df['Store_ID'] = df['Store_ID'].ffill()#Copy the valid Store ID downward
df['Category'] = df['Category'].fillna('Missing')# 3.Replace missing categories with 'Missing' label
df['Temperature_F'] = df['Temperature_F'].interpolate(method='linear')#Estimate missing middle weather values using linear interpolation
df['Sales_USD'] = df['Sales_USD'].fillna(df['Sales_USD'].mean())#Replace missing sales with the average sales
print(df)
```

```python
# 3.Filter and query operations
import pandas as pd
import numpy as np

# Create a dirty dataset with various missing value types[used dictionary]
data = {
    'Date': pd.date_range(start='2026-05-01', periods=6, freq='W'),
    'Store_ID': [101, 101, np.nan, 101, 101, np.nan],
    'Category': ['Electronics', np.nan, 'Electronics', 'Electronics', 'Clothing', np.nan],
    'Temperature_F': [65.0, np.nan, np.nan, 72.0, 75.0, 78.0],
    'Sales_USD': [12000, 14000, 11000, np.nan, 18000, np.nan],
    'Bonus_Payout': [np.nan, np.nan, np.nan, np.nan, np.nan, np.nan]
}

df = pd.DataFrame(data) #data is a dictionary
print("Original DataFrame:")
print(df)
print('\n')
high_sales = df[df['Sales_USD'] > 12000] #Filters all sales >12000
electronics_high_sales = df[(df['Category'] == 'Electronics') & (df['Sales_USD'] > 12000)] #Multiple conditions
specific_data = df[(df['Category'] == 'Clothing') | (df['Temperature_F'] > 75)] #Same here
#Query does same thing as filtering but in more readable prompts
result = df.query("Category == 'Electronics' and Sales_USD > 12000")
sales_info = df.loc[df['Sales_USD'] > 12000, ['Date', 'Sales_USD']] #Output desired columns based on condition
first_chunk = df.iloc[0:3, 0:2] #Integer based indexing
print(high_sales)
print('\n')
print(electronics_high_sales)
print('\n')
print(specific_data)
print('\n')
print(result)
print('\n')
print(sales_info)
print('\n')
print(first_chunk)
```

```python
# 4.Group by operations and aggregations
import pandas as pd
import numpy as np

# Sample dataset with multiple stores and categories
data = {
    'Store_ID': [101, 101, 101, 102, 102, 102],
    'Category': ['Electronics', 'Electronics', 'Clothing', 'Electronics', 'Clothing', 'Clothing'],
    'Sales_USD': [12000, 14000, 18000, 15000, 9000, 22000],
    'Transactions': [80, 95, 120, 110, 45, 150]
}
df = pd.DataFrame(data)
total_sales = df.groupby('Store_ID')['Sales_USD'].sum() #Groups by Store_ID and sums columns
avg_by_category = df.groupby('Category').mean(numeric_only=True) #Calculates mean for each category
sales_stats = df.groupby('Store_ID')['Sales_USD'].agg(['sum', 'mean', 'max']) #Multiple operations simultaneously
print(df)
print('\n')
print(total_sales)
print('\n')
print(avg_by_category)
print('\n')
print(sales_stats)
```

```python
# 5.Merge and join different datasets
import pandas as pd
#Transaction Ledger(Sales data)
df1 = pd.DataFrame({
    'Transaction_ID': [1001, 1002, 1003, 1004],
    'Customer_ID': ['C301', 'C302', 'C301', 'C304'],
    'Product': ['Laptop', 'Headphones', 'Mouse', 'Keyboard'],
    'Amount_INR': [75000, 3500, 1500, 2500]
})

#Customer Profiles(Metadata)
df2 = pd.DataFrame({
    'Customer_ID': ['C301', 'C302', 'C303'],
    'Name': ['Aarav', 'Ananya', 'Vivaan'],
    'City': ['New Delhi', 'Mumbai', 'Bangalore']
})
print("Transaction Ledger:")
print(df1)
print("\nCustomer Profiles:")
print(df2)
inner_result = pd.merge(df1, df2, on='Customer_ID', how='inner')#Keeps customers who made a purchase and have a profile(only)[Inner Join]
print("\nInner Join Result:")
print(inner_result)
left_result = pd.merge(df1, df2, on='Customer_ID', how='left')#Keeps all transactions, fills missing customer profiles with NaN
print("\nLeft Join Result:")
print(left_result)
outer_result = pd.merge(df1, df2, on='Customer_ID', how='outer')#Keeps everything(all sales and all customers, matching where possible)
print("\nOuter Join Result:")
print(outer_result)
```
