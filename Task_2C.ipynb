```python
#1.Line plots with multiple series
import matplotlib.pyplot as plt
import pandas as pd

#DataFrame
data = {
    'Week': ['Week 1', 'Week 2', 'Week 3', 'Week 4'],
    'Store_A_Sales': [12000, 14000, 11000, 15000],
    'Store_B_Sales': [9000, 11500, 13000, 14500]
}
df = pd.DataFrame(data)

#Initialize canvas figure size
plt.figure(figsize=(8, 5))

#Plot series lines
plt.plot(df['Week'], df['Store_A_Sales'], label='Store A', color='royalblue', marker='o', linewidth=2)
plt.plot(df['Week'], df['Store_B_Sales'], label='Store B', color='darkorange', marker='s', linestyle='--')

#Add chart elements
plt.title('Weekly Sales Comparison', fontsize=14, fontweight='bold')
plt.xlabel('Timeline', fontsize=11)
plt.ylabel('Revenue (INR)', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)  #Add soft reference background grid lines

#Display the legend
plt.legend(loc='upper left')

#Render
plt.tight_layout()
plt.show()
```

```python
# 2.Scatter plots with color coding
import matplotlib.pyplot as plt
import pandas as pd

#Sample Data
data = {
    'Ad_Spend': [1000, 1500, 2000, 2500, 3000, 3500],
    'Sales_Units': [150, 250, 400, 350, 600, 550],
    'Region': ['North', 'South', 'North', 'South', 'North', 'South']
}
df = pd.DataFrame(data)

# Create a fixed color mapping dictionary for discrete groups
color_map = {'North': 'royalblue', 'South': 'darkorange'}

plt.figure(figsize=(8, 5))

#Build an automated segmented legend
for region, group in df.groupby('Region'):
    plt.scatter(
        group['Ad_Spend'],
        group['Sales_Units'],
        label=region,
        color=color_map[region],
        s=120,
        edgecolors='black', #Add a border around markers
        alpha=0.8
    )

plt.title('Regional Performance Breakdown', fontsize=12, fontweight='bold')
plt.xlabel('Advertising Spend (INR)')
plt.ylabel('Units Sold')
plt.legend(title='Regions')
plt.grid(True, linestyle=':', alpha=0.5)
plt.show()
```

```python
# 3.Histograms and distribution plots
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

#Sample Data
delivery_hours = np.random.exponential(scale=24, size=300)
df_delivery = pd.DataFrame({'Hours': delivery_hours})

plt.figure(figsize=(8, 5))

#Plot normalized histogram (density=True scales total bar area to 1)
plt.hist(df_delivery['Hours'], bins=25, density=True, alpha=0.5, color='purple', edgecolor='white')

#Overlay pandas smooth KDE curve on the same plot axis
df_delivery['Hours'].plot(kind='kde', color='darkviolet', linewidth=2.5, label='KDE Trend Line')

plt.title('Delivery Time Probability Density', fontsize=12, fontweight='bold')
plt.xlabel('Hours taken for Delivery')
plt.ylabel('Probability Density')
plt.xlim(0, 100) #Limits long tail outliers for visibility
plt.legend()
plt.show()
```

```python
# 4.Subplots and figure customization
import matplotlib.pyplot as plt
import numpy as np

#Geometric progression lines
x = np.linspace(0, 10, 100)

#Create canvas setup with 2 rows, 2 columns
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))

#Plot individual series on specific coordinate grid cells
axes[0, 0].plot(x, x, color='royalblue')
axes[0, 0].set_title('Linear Trend')

axes[0, 1].plot(x, x**2, color='darkorange')
axes[0, 1].set_title('Quadratic Curve')

axes[1, 0].plot(x, np.sin(x), color='teal')
axes[1, 0].set_title('Sine Wave Pattern')

axes[1, 1].plot(x, np.exp(x/2), color='crimson')
axes[1, 1].set_title('Exponential Growth')

#Global custom refinement parameters
for ax in axes.flat:
    ax.set_xlabel('X Axis Axis Label')
    ax.set_ylabel('Y Axis Value')
    ax.grid(True, linestyle=':', alpha=0.6)

#Add window header title and prevent text overlapping
plt.suptitle('Master Performance Overview Matrix', fontsize=14, fontweight='bold')
plt.tight_layout() #Automatically resizes layout spacing boundaries cleanly
plt.show()
```
