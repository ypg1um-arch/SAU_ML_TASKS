```python
# 1.Remove duplicates and outliers
import pandas as pd
import numpy as np
# Sample Data
data = {
    'Customer_ID': ['C01', 'C02', 'C01', 'C03', 'C04'],
    'Product': ['Shoes', 'Hat', 'Shoes', 'Watch', 'Shirt'],
    'Amount_INR': [3500, 1200, 3500, 999999, 2500]
}
df = pd.DataFrame(data)
print(df)
print('\n')
df = df.drop_duplicates()#Clean duplicate rows
Q1 = df['Amount_INR'].quantile(0.25)
Q3 = df['Amount_INR'].quantile(0.75)
IQR = Q3 - Q1

df_clean = df[(df['Amount_INR'] >= (Q1 - 1.5 * IQR)) & (df['Amount_INR'] <= (Q3 + 1.5 * IQR))]#define range of accepted values(using inter quartile range)
print(df_clean)
```

```python
#2.Convert data types appropriately
import pandas as pd
import numpy as np

# Sample Data
data = {
    'Store_ID': [101, 102, 103],                    # Needs to be a string
    'Date_Str': ['2026-05-01', '2026-05-02', 'NaT'], # Needs to be actual datetime
    'Revenue': ['12000', '15500', 'Missing Data'],  # Needs to be numeric (with a clean-up)
    'Rating': [4, 5, 3]                             # Needs to be a float
}
df = pd.DataFrame(data)

# Apply conversions
df['Store_ID'] = df['Store_ID'].astype(str)
df['Date_Str'] = pd.to_datetime(df['Date_Str'], errors='coerce')#coerce means that return unchanged value if found erroneous value
df['Revenue']  = pd.to_numeric(df['Revenue'], errors='coerce')
df['Rating']   = df['Rating'].astype(float)

print(df)
print("\nFinal Column Types:")
print(df.dtypes)
```

```python
# 3.Create derived columns[new data fields from existing attributes using mathematical, conditional, or text-based transformations]
import pandas as pd
import numpy as np
#Sample data
data = {
    'Order_Date': pd.to_datetime(['2026-05-23', '2026-05-24', '2026-05-25']),
    'Product_Ref': ['  p001  ', 'P002', 'p003 '],
    'Units_Sold': [10, 5, 20],
    'Unit_Cost_INR': [500, 1200, 350]
}
df = pd.DataFrame(data)

df['Total_Cost_INR'] = df['Units_Sold'] * df['Unit_Cost_INR']

df['Clean_Ref'] = df['Product_Ref'].str.strip().str.upper()

df['Day_Name'] = df['Order_Date'].dt.day_name()

rules = [
    (df['Total_Cost_INR'] >= 10000),
    (df['Total_Cost_INR'] >= 5000) & (df['Total_Cost_INR'] < 10000)
]
labels = ['Bulk High-Value', 'Medium Order']
df['Order_Type'] = np.select(rules, labels, default='Standard Small')#create labels using rules

print(df)
```

```python
# 4.Handle datetime data
import pandas as pd
# Messy sample data
data = {
    'Order_Str': ['2026-05-01', '2026-05-15', '2026-06-02'],
    'Ship_Str': ['04/05/2026', '16/05/2026', '08/06/2026'],
    'Sales_USD': [500, 1200, 800]
}
df = pd.DataFrame(data)

# 1. Convert columns
df['Order_Date'] = pd.to_datetime(df['Order_Str'])
df['Ship_Date'] = pd.to_datetime(df['Ship_Str'], format='%d/%m/%Y')

# 2. Extract properties & math
df['Weekday'] = df['Order_Date'].dt.day_name()
df['Days_Between'] = (df['Ship_Date'] - df['Order_Date']).dt.days

print(df[['Order_Date', 'Ship_Date', 'Weekday', 'Days_Between']])
```
